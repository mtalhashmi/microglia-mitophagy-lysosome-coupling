# Notebook 05: Overview figure (Figure 1)

**Goal.** Draw both cohorts in one map coloured by cohort, state and age. **For display only**: no statistic anywhere uses this map.

**Cell 1.** Loads the tools and the shared settings. **sys.path.insert** tells Python where the project code lives; **mlc.config** gives every notebook the same folders (PROC = processed data, RES = results tables, FIG = figures) and the profile (*sample* = built-in simulated data, *real* = your downloads). It also tells pandas to print wide tables on one screen. This cell is identical in every notebook.

```python
import sys; sys.path.insert(0, "../src")
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from mlc.config import *
from mlc import genesets as G
pd.set_option("display.width", 150); pd.set_option("display.max_columns", 20)
print("profile:", PROFILE, "| cohorts:", cohorts())
```

**Cell 2.** Joins the cohorts (shared genes only), repeats normalisation and PCA with each sample's average removed (so batch offsets do not dominate the picture), and draws a 2-D t-SNE map of 3,000 random cells. **Important:** removing offsets can change correlations, which is why this map is never used for coupling statistics.

*You should see:* Three panels. Cohorts should be intermixed; states form groups; age shows a gradient along the DAM direction.

```python
from mlc.dataset import Dataset, concat
from mlc import proc, plots
parts = [Dataset.load(PROC / f"{c}_microglia") for c in cohorts()["mouse"]]
A = concat(parts, join="inner")
proc.normalize(A); proc.pca(A, n_comps=30, center_by="sample_id")
idx, emb = proc.display_embedding(A, max_cells=3000)
plots.fig_overview(emb, A.obs.iloc[idx]); plt.show()
```
